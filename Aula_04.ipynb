```python
import pandas as pd
from scipy.spatial.distance import cosine


def busca_semantica(pergunta, chunks, top_k=3):

  emb_pergunta = get_embeddings(pergunta)

  resultados = []
  for idx, chunk in enumerate(chunks):

    emb_chunk = get_embeddings(chunk)

    dist_cosseno = cosine(emb_pergunta, emb_chunk)
    sim_cosseno = 1 - dist_cosseno
    dist_euclidiana = np.linalg.norm(
        np.array(emb_pergunta) - np.array(emb_chunk)
    )

    resultados.append({
        "ID": idx,
        "Trecho": chunk[:80] + "..." if len(chunk) > 80 else chunk,
        "Similaridade Cosseno": round(sim_cosseno, 4),
        "Distância Euclidiana": round(dist_euclidiana, 4),
        "Distância Cosseno": round(dist_cosseno, 4),
    })

  df = pd.DataFrame(resultados)
  df = df.sort_values(by="Similaridade Cosseno", ascending=False).reset_index(
      drop=True
  )

  return df.head(top_k)
```

```python
import glob
import json
import os
import re
import time
from pathlib import Path
from google.colab import userdata
import numpy as np
import pandas as pd
import requests

from langchain_text_splitters import (
    CharacterTextSplitter,
    MarkdownHeaderTextSplitter,
    RecursiveCharacterTextSplitter,
)

API_KEY = userdata.get("OPENROUTER_API_KEY")

URL_DA_SUA_API = "https://openrouter.ai/api/v1/embeddings"
SEUS_HEADERS = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json",
}

PASTA_MARKDOWN = Path("./")
PASTA_SAIDA = Path("./results")


def get_embeddings(text):
  max_retries = 5
  delay = 3

  if len(text) > 20000:
    text = text[:20000]

  for attempt in range(max_retries):
    try:
      response = requests.post(
          URL_DA_SUA_API,
          headers=SEUS_HEADERS,
          json={"model": "openai/text-embedding-3-small", "input": text},
      )
      data = response.json()

      if "data" in data and len(data["data"]) > 0:
        return data["data"][0]["embedding"]

      data_str = str(data)
      if "maximum context length" in data_str or "400" in data_str:
        text = text[: len(text) // 2]
        print(
            f"  ⚠️ Chunk excedeu o limite. Ajustado para {len(text)} caracteres."
        )
        continue

      print(
          f"  ⚠️ Aviso da API (tentativa {attempt + 1}/{max_retries}): {data}."
          f" Aguardando {delay}s..."
      )
    except Exception as e:
      print(
          f"  ⚠️ Falha de conexão (tentativa {attempt + 1}/{max_retries}):"
          f" {e}. Aguardando {delay}s..."
      )

    time.sleep(delay)
    delay *= 2

  raise RuntimeError("❌ A API de embeddings falhou repetidamente.")


def processar_documento(doc_id, nome_pdf, conteudo_md):
  experimentos_resumo = []

  testes_config = {
      1: (
          "fixed",
          200,
          0,
          CharacterTextSplitter(chunk_size=200, chunk_overlap=0, separator=""),
      ),
      2: (
          "fixed",
          500,
          0,
          CharacterTextSplitter(chunk_size=500, chunk_overlap=0, separator=""),
      ),
      3: (
          "fixed",
          1000,
          0,
          CharacterTextSplitter(
              chunk_size=1000, chunk_overlap=0, separator=""
          ),
      ),
      4: (
          "fixed",
          2000,
          0,
          CharacterTextSplitter(
              chunk_size=2000, chunk_overlap=0, separator=""
          ),
      ),
      5: (
          "fixed_with_overlap",
          500,
          50,
          CharacterTextSplitter(
              chunk_size=500, chunk_overlap=50, separator=""
          ),
      ),
      6: (
          "fixed_with_overlap",
          500,
          200,
          CharacterTextSplitter(
              chunk_size=500, chunk_overlap=200, separator=""
          ),
      ),
      7: (
          "paragraph",
          None,
          0,
          RecursiveCharacterTextSplitter(
              chunk_size=500, chunk_overlap=0, separators=["\n\n"]
          ),
      ),
      9: (
          "recursive",
          500,
          50,
          RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50),
      ),
  }

  for test_id in range(1, 11):
    chunks = []
    metadata_list = []
    strategy_name = ""
    c_size = None
    c_overlap = 0

    if test_id in testes_config:
      strategy_name, c_size, c_overlap, splitter = testes_config[test_id]
      chunks = splitter.split_text(conteudo_md)
      metadata_list = [{"char_count": len(c)} for c in chunks]

    elif test_id == 8:
      strategy_name = "sentence_group_3"
      sentencas = [
          s.strip()
          for s in re.split(r"(?<=[.!?])\s+", conteudo_md)
          if s.strip()
      ]
      chunks = [
          " ".join(sentencas[i : i + 3]) for i in range(0, len(sentencas), 3)
      ]
      metadata_list = [
          {"sentencas_no_chunk": min(3, len(sentencas) - i)}
          for i in range(0, len(sentencas), 3)
      ]

    elif test_id == 10:
      strategy_name = "markdown_headers"
      headers_to_split = [
          ("#", "Header 1"),
          ("##", "Header 2"),
          ("###", "Header 3"),
      ]
      md_splitter = MarkdownHeaderTextSplitter(
          headers_to_split_on=headers_to_split
      )
      docs = md_splitter.split_text(conteudo_md)
      chunks = [d.page_content for d in docs]
      metadata_list = [d.metadata for d in docs]

    if not chunks:
      continue

    pasta_teste = PASTA_SAIDA / doc_id / f"test_{test_id:02d}"
    pasta_teste.mkdir(parents=True, exist_ok=True)

    chunks_payload = []
    tamanhos = [len(c) for c in chunks]

    for idx, (chunk_text, meta) in enumerate(zip(chunks, metadata_list)):
      embedding_vector = get_embeddings(chunk_text)

      chunks_payload.append({
          "chunk_id": f"{doc_id}_test{test_id:02d}_chunk{idx+1:03d}",
          "document_id": doc_id,
          "document_name": nome_pdf,
          "test_id": test_id,
          "strategy": strategy_name,
          "chunk_size": c_size,
          "chunk_overlap": c_overlap,
          "text": chunk_text,
          "embedding": (
              embedding_vector.tolist()
              if hasattr(embedding_vector, "tolist")
              else list(embedding_vector)
          ),
          "metadata": meta,
      })

    with open(
        pasta_teste / "chunks_embeddings.json", "w", encoding="utf-8"
    ) as f:
      json.dump(chunks_payload, f, ensure_ascii=False, indent=2)

    experimentos_resumo.append({
        "test_id": test_id,
        "strategy": strategy_name,
        "chunk_size": c_size,
        "chunk_overlap": c_overlap,
        "num_chunks": len(chunks),
        "avg_chunk_size": round(float(np.mean(tamanhos)), 2),
        "min_chunk_size": int(np.min(tamanhos)),
        "max_chunk_size": int(np.max(tamanhos)),
        "embedding_dimension": len(chunks_payload[0]["embedding"]),
    })

  return experimentos_resumo


def executar_pipeline():
  arquivos_md = list(PASTA_MARKDOWN.glob("*.md"))
  print(
      f"🚀 Iniciando/Retomando processamento de {len(arquivos_md)} arquivos"
      " .md...\n"
  )

  summary_global = []

  summary_file = PASTA_SAIDA / "summary.json"
  if summary_file.exists():
    try:
      with open(summary_file, "r", encoding="utf-8") as f:
        summary_global = json.load(f)
    except Exception:
      summary_global = []

  docs_processados = {s["document"] for s in summary_global}

  for arquivo in arquivos_md:
    doc_id = arquivo.stem
    nome_pdf = f"{doc_id}.pdf"

    teste_10_file = PASTA_SAIDA / doc_id / "test_10" / "chunks_embeddings.json"
    if teste_10_file.exists() or nome_pdf in docs_processados:
      print(
          f"  ⏭️ Documento '{doc_id}' já foi totalmente processado. Pulando..."
      )
      continue

    with open(arquivo, "r", encoding="utf-8") as f:
      conteudo_md = f.read()

    pasta_md_out = PASTA_SAIDA / doc_id / "markdown"
    pasta_md_out.mkdir(parents=True, exist_ok=True)
    with open(pasta_md_out / arquivo.name, "w", encoding="utf-8") as f_out:
      f_out.write(conteudo_md)

    experimentos = processar_documento(doc_id, nome_pdf, conteudo_md)

    summary_global = [
        item for item in summary_global if item["document"] != nome_pdf
    ]
    summary_global.append({"document": nome_pdf, "experiments": experimentos})

    with open(summary_file, "w", encoding="utf-8") as f:
      json.dump(summary_global, f, ensure_ascii=False, indent=2)

    print(f"  ✅ Documento '{doc_id}' processado (10 testes ok).")

  print(
      f"\n🎉 Todos os testes foram concluídos! A pasta '{PASTA_SAIDA}' e o"
      " 'summary.json' foram atualizados com sucesso."
  )


executar_pipeline()
```

# Relatório de Avaliação de Estratégias de Chunking com LangChain

## 🛠️ Ferramentas Utilizadas
* **Linguagem & Bibliotecas:** Python, Pandas, SciPy, NumPy, Requests, Re, Pathlib
* **Processamento de Texto:** LangChain Text Splitters (`CharacterTextSplitter`, `RecursiveCharacterTextSplitter`, `MarkdownHeaderTextSplitter`)
* **Modelo de Embeddings:** OpenAI `text-embedding-3-small` (via OpenRouter API)

---

## 🔄 Pipeline de Processamento
1. **Conversão:** PDF → Markdown
2. **Chunking:** Divisão do texto executando as 10 estratégias de Chunking
3. **Vetorização:** Geração de Embeddings para cada chunk
4. **Persistência:** Exportação Estruturada em JSON

---

## 📊 Análise Comparativa dos Experimentos

* **Estratégia com mais chunks:** **Teste 1 (Fixo - 200 caracteres)**, devido à alta fragmentação semântica.
* **Estratégia com menos chunks:** **Teste 4 (Fixo - 2000 caracteres)** e **Teste 10 (Markdown Headers)**, pois agrupam grandes blocos e seções inteiras.
* **Variação de Tamanho:** Os testes baseados em caracteres (1 a 6) mantiveram tamanhos previsíveis. Os testes por estrutura (7, 8 e 10) apresentaram alta variação no tamanho final dos chunks, refletindo a estrutura natural do texto.
* **Tabelas:** Durante a conversão para Markdown, as tabelas são representadas em sintaxe de tabela Markdown (`| coluna |`). Nos testes 1 a 6 (tamanho fixo), linhas de tabelas costumam ser cortadas ao meio. No Teste 10 (Markdown), a integridade estrutural das tabelas é preservada na seção correspondente.
* **Imagens:** Convertidas em marcações/referências no formato Markdown `![imagem](path)` ou descrições, perdendo os vetores de dados visuais nativos.

---

## 🎯 Conclusão e Recomendações para RAG

### ❌ Estratégia a Descartar
* **Teste 1 (Fixo - 200 caracteres):** Deve ser descartado, pois quebra palavras ao meio e destrói o contexto necessário para recuperação semântica.

### 🏆 Estratégias Recomendadas para RAG
1. **Teste 9 (Recursive Splitter):** É a melhor abordagem geral, pois respeita a hierarquia de parágrafos e frases antes de realizar cortes brutos.
2. **Teste 10 (Markdown Splitter):** Excelente para documentos estruturados, pois preserva o contexto temático completo de cada seção.
